```python
df_chuva = df[(df['Precip'] > 0) & (df['Precip'] < 25)]
df_chuva_extrema = df[(df['Precip'] >= 25)]
```

```python
df_chuva = df_chuva[variaveis]
df_chuva_extrema = df_chuva_extrema[variaveis]
```

```python
df_chuva = df_chuva[~(df_chuva == -9999.000000).any(axis=1)]

df_chuva_extrema = df_chuva_extrema[~(df_chuva_extrema == -9999.000000).any(axis=1)]
```

```python
import pandas as pd

# Função para discretizar a direção do vento
def discretizar_direcao_vento(angulo):
    if (angulo >= 0 and angulo < 22.5) or (angulo >= 337.5 and angulo < 360):
        return 'Norte'
    elif angulo >= 22.5 and angulo < 45:
        return 'Norte-Nordeste'
    elif angulo >= 45 and angulo < 67.5:
        return 'Nordeste'
    elif angulo >= 67.5 and angulo < 90:
        return 'Leste-Nordeste'
    elif angulo >= 90 and angulo < 112.5:
        return 'Leste'
    elif angulo >= 112.5 and angulo < 135:
        return 'Sul-Leste'
    elif angulo >= 135 and angulo < 157.5:
        return 'Sudeste'
    elif angulo >= 157.5 and angulo < 180:
        return 'Sul-Sudeste'
    elif angulo >= 180 and angulo < 202.5:
        return 'Sul'
    elif angulo >= 202.5 and angulo < 225:
        return 'Sudoeste'
    elif angulo >= 225 and angulo < 247.5:
        return 'Oeste-Sudoeste'
    elif angulo >= 247.5 and angulo < 270:
        return 'Oeste'
    elif angulo >= 270 and angulo < 292.5:
        return 'Noroeste'
    elif angulo >= 292.5 and angulo < 315:
        return 'Noroeste-Norte'
    elif angulo >= 315 and angulo < 337.5:
        return 'Norte-Noroeste'
    
```

```python

# Filtrar apenas os dados da estação "FORTE_COPACABANA"
df_COPACABANA = df_chuva[df_chuva['ESTACAO'] == 'FORTE_COPACABANA']

# Contar a quantidade de cada categoria na coluna de direção do vento
contagem_direcao_vento = df_COPACABANA['DIRECAO DO VENTO PRIMEIRA HORA ANTERIOR'].value_counts()

# Ordenar os valores de forma decrescente
contagem_ordenada = contagem_direcao_vento.sort_values(ascending=False)

# Calcular a porcentagem acumulada
porcentagem_acumulada = contagem_ordenada.cumsum() / contagem_ordenada.sum() * 100

# Criar o gráfico de Pareto
fig, ax1 = plt.subplots(figsize=(12, 6))

# Barras (Frequência)
ax1.bar(contagem_ordenada.index, contagem_ordenada, color='skyblue', alpha=0.6, label='Frequência')
ax1.set_xlabel('Direção do Vento', fontsize=12)
ax1.set_ylabel('Frequência', color='b', fontsize=12)
ax1.tick_params(axis='y', labelcolor='b')

# Criar eixo secundário para a porcentagem acumulada
ax2 = ax1.twinx()
ax2.plot(contagem_ordenada.index, porcentagem_acumulada, color='r', marker='o', linestyle='-', linewidth=2, label='Porcentagem Acumulada (%)')
ax2.set_ylabel('Porcentagem Acumulada (%)', color='r', fontsize=12)
ax2.tick_params(axis='y', labelcolor='r')

# Adicionar rótulos na linha da porcentagem acumulada
for i, (cat, pct) in enumerate(zip(contagem_ordenada.index, porcentagem_acumulada)):
    ax2.text(i, pct + 1, f"{pct:.1f}%", ha='center', va='bottom', fontsize=10, color='red')

# Rotacionar rótulos do eixo X corretamente
ax1.set_xticks(range(len(contagem_ordenada.index)))
ax1.set_xticklabels(contagem_ordenada.index, rotation=45, ha='right')

# Ajustar título e layout
plt.title('Gráfico de Pareto - Direção do Vento na hora anterior a medição de chuva (FORTE COPACABANA)', fontsize=14)
plt.tight_layout()
plt.show()




```
