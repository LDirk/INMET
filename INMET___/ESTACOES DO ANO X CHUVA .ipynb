```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np 

from IPython.display import display
```

```python
# Caminho para o arquivo Parquet
caminho_parquet = r"C:\Users\Pichau\INMET\TODAS_ESTACOES\TODAS_ESTACOES_CONCATENADO.parquet"

# Ler o arquivo Parquet
df = pd.read_parquet(caminho_parquet)
```

```python
df_nao_chuva = df[(df['Precip'] == 0)]

df_chuva = df[(df['Precip'] > 0) & (df['Precip'] < 25)]

df_chuva_extrema = df[(df['Precip'] >= 25)]
```

```python
variaveis = ['Precip', 'estacao_do_ano','Dt_Hr']
```

```python
df_nao_chuva = df_nao_chuva[variaveis]
```

```python
df_nao_chuva.head()
```

```python
contagem_estacoes = df_nao_chuva['estacao_do_ano'].value_counts()  # Contagem absoluta
contagem_relativa = df_nao_chuva['estacao_do_ano'].value_counts(normalize=True)  # Contagem relativa (proporção)

# Exibir os resultados
print("Contagem Absoluta:")
print(contagem_estacoes)

print("\nContagem Relativa:")
print(contagem_relativa)
```

```python
df_chuva = df_chuva[variaveis]
```

```python
contagem_estacoes = df_chuva['estacao_do_ano'].value_counts()  # Contagem absoluta
contagem_relativa = df_chuva['estacao_do_ano'].value_counts(normalize=True)  # Contagem relativa (proporção)

# Exibir os resultados
print("Contagem Absoluta:")
print(contagem_estacoes)

print("\nContagem Relativa:")
print(contagem_relativa)
```

```python
df_chuva_extrema = df_chuva_extrema[variaveis]
```

```python
contagem_estacoes = df_chuva_extrema['estacao_do_ano'].value_counts()  # Contagem absoluta
contagem_relativa = df_chuva_extrema['estacao_do_ano'].value_counts(normalize=True)  # Contagem relativa (proporção)

# Exibir os resultados
print("Contagem Absoluta:")
print(contagem_estacoes)

print("\nContagem Relativa:")
print(contagem_relativa)
```

```python
import scipy.stats as stats

contagem_observada = df_chuva_extrema['estacao_do_ano'].value_counts()

contagem_esperada = contagem_observada.sum() / len(contagem_observada)

chi2_stat, p_value = stats.chisquare(contagem_observada, contagem_esperada)

# Exibir os resultados
print(f"Estatística Qui-Quadrado: {chi2_stat:.4f}")

print(f"Valor-p: {p_value:.4f}")

# Verificar significância
if p_value < 0.05:
    print("Rejeitamos H0: Existem diferenças significativas entre as estações.")
    
else:
    print("Não rejeitamos H0: Não há evidências de diferenças significativas entre as estações.")
```
